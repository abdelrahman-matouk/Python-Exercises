# **Excercise 4**

```python

row , col = 3 , 3
matrix = [[0] * 3 for _ in range(3)]
for row in matrix:
    print(row)

# Main diagonal ==> 1s
for i in range(3):
  matrix[i][i] =1
print()  
for row in matrix:
    print(row)

# Anti diagonal ==> 2s   [ (0 , 2) , (1 , 1) , (2 , 0)]
x = 2
for i in range(3): 
  matrix[i][x] = 2
  x = x - 1
print()  
for row in matrix:
    print(row)
print()

print("Final Output:")
matrix[1][1] = 3
for row in matrix:
    print(row)
```

```python
range_size = int(input())
matrix = [[0] * range_size for _ in range(range_size)]
for row in matrix:
    print(row)

# Main diagonal ==> 1s
for i in range(range_size):
  matrix[i][i] =1
print()  
for row in matrix:
    print(row)

# Anti diagonal ==> 2s   [ (0 , 2) , (1 , 1) , (2 , 0)]
x = range_size - 1
for i in range(range_size): 
  matrix[i][x] = 2
  x = x - 1
print()  
for row in matrix:
    print(row)
print()

print("Final Output:")
matrix[range_size//2][range_size//2] = 3
for row in matrix:
    print(row)
```

 **Squares of evens using List Comprehention**

```python
nums = list(map(int,input("Enter Your Nums: ").split()))
print(f" Our Nums : {nums}\n")
Even_Nums = [num for num in nums if num % 2 == 0]
print(f" Our Even_Nums : {Even_Nums}\n")

power_Of_evens = [num ** 2 for num in nums if num % 2 == 0]
print(f" power_Of_evens : {power_Of_evens} ")
```

**No Vowels**

```python
word = input("Enter your word")
print(f"Word is :\n {word}")
NO_vowels = [char for char in word if char not in {'e','o','a','u','i'}]
Str = "".join(NO_vowels)
print(f"The word without vowels is :\n {Str}")
```
